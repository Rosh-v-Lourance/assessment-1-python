```python
# 1. Bank Account Manager
a. Simulate a bank account with deposit, withdrawal, and balance enquiry operations
 b. Use variables and data types to store account information
  c.  Implement conditional statements to check for sufficient balance before withdrawal
d. Use functions to perform deposit, withdrawal, and balance inquiry operations.
```

```python
balance = 20000

def deposit():
    global balance
    amount = float(input("Enter amount to deposit: "))
    balance = balance + amount
    print("Amount deposited successfully")
    print("Balance:", balance)

def withdraw():
    global balance
    amount = float(input("Enter amount to withdraw: "))

    if amount <= balance:
        balance = balance - amount
        print("Withdrawal successful")
        print("Balance:", balance)
    else:
        print("Insufficient balance")

def check_balance():
    print("Your balance is:", balance)


while True:
    print("\n1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        deposit()

    elif choice == "2":
        withdraw()

    elif choice == "3":
        check_balance()

    elif choice == "4":
        print("Thank you!")
        break

    else:
        print("Invalid choice")
```

Implement conditional statements to determine weather conditions d. Use functions to calculate average temperature and humidity, and




```python
region = input("Enter the region: ")

temperature = [30, 32, 31, 29, 33]
humidity = [70, 65, 72, 68, 75]

weather = input("Enter weather condition (sunny/rainy/cloudy): ")

def average_temperature():
    total = sum(temperature)
    average = total / len(temperature)
    return average

def average_humidity():
    total = sum(humidity)
    average = total / len(humidity)
    return average

print("\nWeather Report")
print("Region:", region)
print("Weather:", weather)

if weather == "sunny":
    print("It is a sunny day.")
elif weather == "rainy":
    print("It is a rainy day.")
elif weather == "cloudy":
    print("It is a cloudy day.")
else:
    print("Weather condition is not recognized.")

print("Average Temperature:", average_temperature())
print("Average Humidity:", average_humidity())
```

```python
3.#Password Validator Write a function that checks if a password is strong: at least 8 characters, contains at least one digit, one uppercase letter, and one special character (!@#$%).
```

```python
password = input("Enter a strong password")

upper = False
lower = False
digit = False
specail = False
spl_char = "!@#$%"
for letter in password:
  if letter.isupper():
    upper = True
  elif letter.islower():
    lower = True
  elif letter.isdigit():
    digit = True
  elif letter in spl_char:
    specail = True
if len(password) >= 8 and upper and lower and digit and specail:
  print("Strong password")
else:
  print(" password not strong")
```

```python
# 4 Word Frequency Counter Given a paragraph of text, write a function that returns the 3 most frequently occurring words (ignore case and punctuation).
```

```python
word = input("Enter a word: ")

letter_frequency = {}

for i in word:
    if i in letter_frequency:
        letter_frequency[i] += 1
    else:
        letter_frequency[i] = 1

print(letter_frequency)

# Sort by frequency
sorted_frequency = sorted(letter_frequency.items(), key=lambda x: x[1], reverse=True)

# Get top 3
top_3 = sorted_frequency[:3]

print("Top 3 most frequent letters:", top_3)
```

```python
# 5 FizzBuzz Variant
Print numbers 1 to 50, but replace multiples of 3 with "Fizz", multiples of 5 with "Buzz", multiples of both with "FizzBuzz" — and multiples of 7 with "Bang".
```

```python
for num in range(1,51):
  if num % 3 ==0 and num % 5 ==0:
    print("FizzBuzz")
  elif num % 3 == 0:
    print("Fizz")
  elif num % 5 == 0:
    print("Buzz")
  else:
    print(num)
```
