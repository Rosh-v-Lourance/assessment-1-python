3) find median

```python
def find_median(numbers):
    numbers.sort()
    n = len(numbers)

    if n % 2 == 1:
        median = numbers[n // 2]
    else:
        median = (numbers[n // 2 - 1] + numbers[n // 2]) / 2

    return median


numbers = [10, 5, 20, 15, 25]

result = find_median(numbers)

print("Median:", result)
```

4)  Write a function to return the mod of a list of numbers, if no mod exists, it
should return None. If more than one mod exist, return the mod values as list


```python
def find_mode(numbers):
    frequency = {}

    for i in numbers:
        if i in frequency:
            frequency[i] += 1
        else:
            frequency[i] = 1

    highest = max(frequency.values())

    if highest == 1:
        return None

    modes = []

    for i in frequency:
        if frequency[i] == highest:
            modes.append(i)

    if len(modes) == 1:
        return modes[0]
    else:
        return modes


numbers = [1, 2, 2, 3, 3, 4]

result = find_mode(numbers)

print("Mode:", result)
```

5)Write function to compute a given percentile value (e.g. 90th percentile) of a
list of numbers

```python
def find_percentile(numbers, percentile):
    numbers.sort()

    n = len(numbers)

    position = (percentile / 100) * (n - 1)

    lower = int(position)
    upper = lower + 1

    if upper >= n:
        return numbers[lower]

    value = numbers[lower] + (position - lower) * (numbers[upper] - numbers[lower])

    return value


numbers = [10, 20, 30, 40, 50]

result = find_percentile(numbers, 90)

print("90th Percentile:", result)
```

7)Write function to return any outliers from a given list of numbers, if no
outliers present it should return None


```python
def find_outliers(numbers):
    numbers.sort()

    n = len(numbers)

    q1 = numbers[n // 4]
    q3 = numbers[(3 * n) // 4]

    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outliers = []

    for i in numbers:
        if i < lower or i > upper:
            outliers.append(i)

    if len(outliers) == 0:
        return None
    else:
        return outliers


numbers = [10, 12, 13, 15, 14, 16, 99]

result = find_outliers(numbers)

print("Outliers:", result)
```

8). Write a function to return standard deviation of a list of numbers


```python
import math

def find_standard_deviation(numbers):
    mean = sum(numbers) / len(numbers)

    total = 0

    for i in numbers:
        total = total + (i - mean) ** 2

    variance = total / len(numbers)

    standard_deviation = math.sqrt(variance)

    return standard_deviation


numbers = [10, 20, 30, 40, 50]

result = find_standard_deviation(numbers)

print("Standard Deviation:", result)
```

9). Write function to compute and return Z-score values of a list of numbers

```python
def find_z_scores(numbers):
    mean = sum(numbers) / len(numbers)

    total = 0

    for i in numbers:
        total = total + (i - mean) ** 2

    variance = total / len(numbers)
    standard_deviation = variance ** 0.5

    z_scores = []

    for i in numbers:
        z = (i - mean) / standard_deviation
        z_scores.append(z)

    return z_scores


numbers = [10, 20, 30, 40, 60]

result = find_z_scores(numbers)

print("Z-scores:", result)
```

10)Write a function to compute and return the Pearson correlation coefficient
value of two list of numbers


```python
def pearson_correlation(x, y):
    mean_x = sum(x) / len(x)
    mean_y = sum(y) / len(y)

    numerator = 0
    sum_x = 0
    sum_y = 0

    for i in range(len(x)):
        numerator = numerator + (x[i] - mean_x) * (y[i] - mean_y)
        sum_x = sum_x + (x[i] - mean_x) ** 2
        sum_y = sum_y + (y[i] - mean_y) ** 2

    correlation = numerator / ((sum_x * sum_y) ** 0.5)

    return correlation


x = [10, 20, 30, 40, 50]
y = [15, 25, 35, 45, 55]

result = pearson_correlation(x, y)

print("Pearson Correlation:", result)
```
