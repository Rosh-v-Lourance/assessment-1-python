TASK 8

```python
merged_df = pop.merge(gdp, on='country', how='inner')
merged_df = merged_df.merge(j, on='country', how='inner')
merged_df = merged_df.merge(literacy_rate, on='country', how='inner')

merged_df.head()
```

Task 9

```python
merged_df['Internet Penetration (%)'] = (merged_df['internet_users'] / merged_df['population']) * 100
merged_df['GDP Per Capita'] = merged_df['GDP'] / merged_df['population']

merged_df.head()
```

```python
# Task 10

import sqlite3

conn = sqlite3.connect('country_statistics.db')
merged_df.to_sql('country_statistics', conn, if_exists='replace', index=False)
```

```python
# 11. Display the Top 5 countries by GDP:
q = "SELECT country, GDP FROM country_statistics ORDER BY GDP DESC LIMIT 5"
df= pd.read_sql_query(q, conn)
df
```

```python
# 12. Find the average literacy rate:
q = "SELECT AVG(literacy_rate) FROM country_statistics"
df = pd.read_sql_query(q, conn)
df
```

```python
# 13. Display all countries having literacy rate above 90%:
q = '''SELECT country, literacy_rate
FROM country_statistics
WHERE literacy_rate > 90
ORDER BY literacy_rate DESC '''
df = pd.read_sql_query(q, conn)
df
```

```python
# 14. Find the Top 5 countries with the highest internet penetration:

q = ''' SELECT country, \"Internet Penetration (%)\"
FROM country_statistics
ORDER BY \"Internet Penetration (%)\" DESC LIMIT 5 '''
df = pd.read_sql_query(q, conn)
df
```

```python
# 15. Calculate the total population

q = '''SELECT SUM(population)
FROM country_statistics'''
df = pd.read_sql_query(q, conn)
df
```

```python
# 16. Find the country with the highest GDP per capita:
q = '''SELECT country, \"GDP Per Capita\"
FROM country_statistics
ORDER BY \"GDP Per Capita\" DESC LIMIT 1 '''
df =pd.read_sql_query(q, conn)
df
```
